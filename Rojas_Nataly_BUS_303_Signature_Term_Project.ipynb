```python
dnkn2 = dnkn1.copy()
```

```python
dnkn2['Cholesterol (mg)'] = dnkn2['Cholesterol (mg)'].replace('<5', '5')
print(" '<5' replaced with '5' in 'Cholesterol (mg)' column.")
```

```python
for col in dnkn2.select_dtypes(include='object').columns:
    dnkn2[col] = dnkn2[col].str.replace('\xa0', '', regex=False).str.strip()
print(" '\xa0' removed and whitespace stripped from object columns.")
```

```python
for col in dnkn2.columns:
    if dnkn2[col].dtype == 'object':
        print('\n', col)
        print(dnkn2[col].unique()[:50])
```

```python
#In Deliverable Two, the primary objective was to clean and prepare the Dunkin' Donuts nutritional dataset for analysis. Initially, the dataset contained several issues that could impede accurate statistical interpretation. These included the presence of duplicate records, which could lead to overcounting observations and special characters within numerical columns, such as "<5" and "\xa0" that prevented proper data type conversion and numerical operations. Additionally, some rows represented food categories rather than specific beverage products and there were empty string entries that needed to be treated as missing values. To address these issues, a systematic cleaning process was undertaken. Duplicate records were identified and removed. Special characters were systematically cleaned from relevant columns and empty strings were converted into NaN (Not a Number) values to be properly handled as missing data. Rows identified as food categories were filtered out to ensure that the dataset exclusively contained beverage items. Finally, all relevant columns were converted to appropriate numeric data types, and the entire dataset was validated through summary statistics, data type checks, and shape verification to ensure it was clean, consistent, and ready for further analysis.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
portfolio = pd.DataFrame({
 "Company": ["Starbucks", "Dunkin' Donuts"],
 "Beverage_combinations": [len(sbux), len(dnkn4)] # actual counts from your data
})
# Define the official colors for Starbucks and Dunkin' Donuts
starbucks_green = "#00704A"
dunkin_pink = "#FF5E9E" # Dunkin’ bright pink
colors = [starbucks_green, dunkin_pink]
```

```python
ax = portfolio.plot(kind= 'bar' , x='Company', y='Beverage_combinations', color=colors, legend = False)
plt.title('Beverage Combination Portfolio Size')
plt.xlabel('Company')
plt.ylabel('Number of Beverages')
for i in ax.containers:
    ax.bar_label(i)
plt.tight_layout()
plt.show()
```

```python
avg_calories = pd.DataFrame({
 "Company": ["Starbucks", "Dunkin' Donuts"],
 "Average_Calories": [
 sbux["Calories"].mean(),
 dnkn4["Calories"].mean()
 ]
})
# Define the official colors for Starbucks and Dunkin' Donuts
starbucks_green = "#00704A"
dunkin_pink = "#FF5E9E"
# Create a Horizontal Bar chart
ax = avg_calories.plot(
 kind="bar",
 x="Company",
 y="Average_Calories",
 color=[starbucks_green, dunkin_pink], figsize=(8,5), legend = False)
plt.title('Average Beverage Calories')
plt.ylabel('Calories')
for p in ax.patches:
    ax.annotate(str(round(p.get_height(), 1)),
                xy=(p.get_x() + p.get_width() / 2., p.get_height()),
                ha = 'center', va = 'center', xytext = (0,10), 
                textcoords = 'offset points')
plt.show()
```

```python
# Calculate the average sugar for each menu.
avg_sugar = pd.DataFrame({
 "Company": ["Starbucks", "Dunkin' Donuts"],
 "Average_Sugar": [
 sbux["Sugar_g"].mean(),
 dnkn4["Sugars (g)"].mean()
 ]
})
# Create a Horizontal Bar chart
ax = avg_sugar.plot(
 kind="bar",
 x="Company",
 y="Average_Sugar",
 color=[starbucks_green, dunkin_pink],
 figsize=(8,5),
 legend=False
)
# Add Titles & labels
plt.title("Average Beverage Sugar Content", fontsize=14, fontweight="bold")
plt.ylabel("Sugar (g)", fontsize=12)
for p in ax.patches:
 ax.annotate(
 f"{p.get_height():.1f}",
 (p.get_x()+p.get_width()/2., p.get_height()),
 ha="center",
 va="bottom",
 fontsize=11,
 fontweight="bold"
 )
for p in ax.patches:
    ax.annotate(f'{p.get_height():.1f}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha = 'center', va = 'bottom', fontsize = 11, fontweight='bold')
plt.show()
```

```python
# Calculate average protein for each menu.
protein = pd.DataFrame({
 "Company": ["Starbucks", "Dunkin' Donuts"],
 "Average_Protein": [
 sbux["Protein_g"].mean(),
 dnkn4["Protein (g)"].mean()

]
})
# Create a Horizontal Bar chart
ax = protein.plot(
 kind="bar",
 x="Company",
 y="Average_Protein",
 color=[starbucks_green, dunkin_pink],
 figsize=(8,5),
 legend=False )
for i, v in enumerate(protein['Average_Protein']):
    ax.text(i, v + 0.1, str(round(v, 2)),
    ha = 'center', va = 'bottom')
ax.set_xlabel("Company")
ax.set_ylabel("Average Beverage Protein (g)")
ax.set_title("Average Beverage Protein Content")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define a threshold for what is acceptably healthy in terms of calories and sugar for Starbucks.
sbux_healthy = len(
 sbux[
 (sbux["Calories"] < 100) &
 (sbux["Sugar_g"] < 15)
 ]
)
# Define a threshold for what is acceptably healthy in terms of calories and sugar for Dunkin' Donuts
dnkn_healthy = len(
 dnkn4[
 (dnkn4["Calories"] < 100) &
 (dnkn4["Sugars (g)"] < 15)
 ]
)
# Create a dataframe of the number of healthy drink combinations for both menus.
healthy = pd.DataFrame({
 "Company":["Starbucks","Dunkin' Donuts"],
 "Healthy_Beverages":[
 sbux_healthy,
 dnkn_healthy
 ]
})
bars = plt.bar(healthy['Company'], healthy['Healthy_Beverages'], color=['#006241' , '#FF69B4'])
for bar in bars:
    yval= bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval+2, yval, ha='center', fontweight = 'bold')
plt.ylabel('Number of Beverages')
plt.xlabel('Company')
plt.title('Healthier Beverage Portfolio')
plt.show()
```
